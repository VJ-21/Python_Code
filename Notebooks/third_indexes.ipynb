```python
import pandas as pd
# Indexes-How to set,reset and use indexes
```

```python
df=pd.read_csv('D:/Python_Practice/Notebooks/banklist.csv',index_col='CERT')
```

```python
df
```

```python
# setting index to a column
# indexing is the process of setting a column as the index of the data frame. This allows us to access the rows of the data frame based on the values in that column. We can set the index to a column using the set_index() method. The set_index() method takes the name of the column as an argument and returns a new data frame with the specified column as the index.
df.set_index('CERT',inplace=True)
```

```python
df.index
```

```python
df.loc[35386]
# Accessing rows using the index created on column 'CERT' using the loc method. The loc method takes the value of the index as an argument and returns the row corresponding to that index value.
```

```python
# reset_index() method is used to reset the index of the data frame. This method takes no arguments and returns a new data frame with the index reset to the default integer index. The original index is added as a new column in the data frame.
df.reset_index(inplace=True)
```

```python
df
```

```python
df.loc[[35386],['Bank Name','City','ST']]

# After setting the index at initial dataframe itself, we can access the rows using the loc method. The loc method takes the value of the index as an argument and returns the row corresponding to that index value. We can also access specific columns of the row by passing a list of column names as the second argument to the loc method.
```
