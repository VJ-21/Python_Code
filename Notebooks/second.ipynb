```python
# Creating dictionary and converting it into data frame to understand the data frame structure and its attributes
# DataFrame and series Basics - Selecting Rows and Columns
import pandas as pd
```

```python
people={
    "Name":['John','Smith','Alice','Bob'],
    "age":[25,30,22,28],
    "email":['John@example.com','Smith@example.com','Alice@example.com','Bob@example.com']
}
```

```python
df=pd.DataFrame(people)
```

```python
df
```

```python
df['email']
# to access a particular column of the data frame or df.email
```

```python
type(df['email'])
# printing the type of the column to understand the data type of the column
# series is a one-dimensional array that can hold any data type, while a data frame is a two-dimensional table with rows and columns.
```

```python
# Accessing multiple columns of the data frame passing a list of column names to the data frame. This will return a new data frame with only the specified columns.
df[['Name','email']]
```

```python
# loc iloc  iloc are used to access the rows of the data frame based on their index position. loc is used to access rows by label, while iloc is used to access rows by integer position.
# [[row_index1,row_index2,...], [col_index1,col_index2,...]]  to access multiple rows and columns of the data frame based on their index position. This will return a new data frame with only the specified rows and columns.
df.iloc[[0,1], [0]]
```

```python
df.loc[[0,1],['email','age']]
```
